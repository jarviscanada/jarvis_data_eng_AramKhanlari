```python
from pyspark.sql.functions import year, month
df = df.withColumn(
    "yyyymm",
    year("invoice_date") * 100 + month("invoice_date")
)
invoice_amounts = (
    df
    .groupBy("invoice_no", "yyyymm").sum("invoice_total")
    .withColumnRenamed("sum(invoice_total)", "invoice_total")
)
df =df.withColumn("invoice_status",col("invoice_total") >= 0)
#display(df)
display(invoice_amounts)
#display(invoice_status)
```

```python
from pyspark.sql.functions import countDistinct
monthly_canceled = (
    df
    .filter(col("invoice_status") == False)
    .groupBy("yyyymm")
    .agg(
        countDistinct("invoice_no").alias("canceled_orders")
    )
)
```

```python
from pyspark.sql.functions import count
monthly_total = (
    invoice_amounts
    .groupBy("yyyymm")
    .agg(
        count("invoice_no").alias("total_invoices")
    )
)
```

```python
monthly_summary = (
    monthly_total
    .join(monthly_canceled, on="yyyymm", how="left")
    .fillna(0) 
)
display(monthly_summary)
```

# Monthly Sales Growth


```python
monthly_sales = (
    df
    .groupBy("yyyymm")
    .agg(
        spark_sum("invoice_total").alias("total_sales")
    )
    .orderBy("yyyymm")
)
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import col, lag
window_spec = Window.orderBy("yyyymm")
monthly_sales = monthly_sales.withColumn(
    "prev_sales",
    lag("total_sales").over(window_spec)
)
monthly_growth = monthly_sales.withColumn(
    "monthly_growth_pct",
    ((col("total_sales") - col("prev_sales")) / col("prev_sales")) * 100
)
display(monthly_growth)
```

```python
from pyspark.sql.functions import min

first_purchase = (
    df
    .groupBy("customer_id")
    .agg(
        min("yyyymm").alias("first_yyyymm")
    )
)

# Join df with first_purchase to enable comparison
df_with_first = df.join(first_purchase, on="customer_id", how="left")

# Now we can compare columns from the joined DataFrame
df_with_first = df_with_first.withColumn("is_new", col("yyyymm") == col("first_yyyymm"))

monthly_new = (
    df_with_first.filter(col("is_new"))
    .groupBy("yyyymm")
    .agg(
        countDistinct("customer_id").alias("new_users")
    )
)

monthly_existing = (
    df_with_first.filter(~col("is_new"))
    .groupBy("yyyymm")
    .agg(
        countDistinct("customer_id").alias("existing_users")
    )
)

monthly_users = monthly_new.join(monthly_existing, on="yyyymm", how="outer")
display(monthly_users)
```

## Finding RFM

RFM is a method used for analyzing customer value. It is commonly used in database marketing and direct marketing and has received particular attention in the retail and professional services industries. ([wikipedia](https://en.wikipedia.org/wiki/RFM_(market_research)))

Optional Reading: [Making Your Database Pay Off Using Recency Frequency and Monetary Analysis](http://www.dbmarketing.com/2010/03/making-your-database-pay-off-using-recency-frequency-and-monetary-analysis/)


RFM stands for three dimensions:

- Recency – How recently did the customer purchase?

- Frequency – How often do they purchase?

- Monetary Value – How much do they spend?

<!-- Note: To simplify the problem, let's keep all placed and canceled orders.


**Sample RFM table**

![](https://i.imgur.com/sXFIg6u.jpg) -->

```python
from pyspark.sql.functions import max, datediff, col ,lit
df_recency =(
    df
    .groupby("customer_id")
    .agg(
        max("invoice_date").alias("LastPurchaseDate")
    )
)
recent_date =(
    df_recency
    .agg(max("LastPurchaseDate").alias("recent_date"))
    .collect()[0]["recent_date"]
)
df_recency = df_recency.withColumn(
    "Recency",
    datediff(lit(recent_date), col("LastPurchaseDate"))
)
display(df_recency)
```

```python
from pyspark.sql.functions import count
frequency_df = (
    df
    .dropDuplicates()
    .groupBy("customer_id")
    .agg(
        count("invoice_no").alias("Frequency")
    )
)
display(frequency_df)
```

```python
from pyspark.sql.functions import sum as spark_sum

monetary_df = (
    df
    .groupBy("customer_id")
    .agg(
        spark_sum("invoice_total").alias("Monetary")
    )
)
display(monetary_df)
```

```python
rfm_df = (
    df_recency
    .join(frequency_df, on='customer_id', how ="inner")
    .join(monetary_df, on='customer_id', how ="inner")
    .drop("LastPurchaseDate")
)

display(rfm_df)
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import ntile
recency_window = Window.orderBy(col("Recency").asc())
rfm_scored = rfm_df.withColumn(
    "RecencyScore",
    (6 - ntile(5).over(recency_window)) 
)
frequency_window = Window.orderBy(col("Frequency").desc())
rfm_scored = rfm_scored.withColumn(
    "FrequencyScore",
    (ntile(5).over(frequency_window)) 
)
monetary_window = Window.orderBy(col("Monetary").desc())
rfm_scored = rfm_scored.withColumn(
    "MonetaryScore",
    (ntile(5).over(monetary_window)) 
)
display(rfm_scored)
```

```python
rfm_scored = rfm_scored.withColumn(
    "RFM_SCORE",
    (rfm_scored['RecencyScore'] * 100 + 
     rfm_scored['FrequencyScore'] *10 + 
     rfm_scored['MonetaryScore'])
)
display(rfm_scored)
```

```python
from pyspark.sql.functions import when
rfm_segmented = (
    rfm_scored
    .withColumn(
        "Segment",
        when((col("RecencyScore").between(1, 2)) & (col("FrequencyScore").between(1, 2)), "Hibernating")
        .when((col("RecencyScore").between(1, 2)) & (col("FrequencyScore").between(3, 4)), "At Risk")
        .when((col("RecencyScore").between(1, 2)) & (col("FrequencyScore") == 5), "Can't Lose")

        .when((col("RecencyScore") == 3) & (col("FrequencyScore").between(1, 2)), "About to Sleep")
        .when((col("RecencyScore") == 3) & (col("FrequencyScore") == 3), "Need Attention")
        .when((col("RecencyScore").between(3, 4)) & (col("FrequencyScore").between(4, 5)), "Loyal Customers")

        .when((col("RecencyScore") == 4) & (col("FrequencyScore") == 1), "Promising")
        .when((col("RecencyScore") == 5) & (col("FrequencyScore") == 1), "New Customers")

        .when((col("RecencyScore").between(4, 5)) & (col("FrequencyScore").between(2, 3)), "Potential Loyalists")
        .when((col("RecencyScore") == 5) & (col("FrequencyScore").between(4, 5)), "Champions")

        .otherwise("Other")
    )
)
display(rfm_segmented.select(
    "customer_id",
    "RecencyScore",
    "FrequencyScore",
    "MonetaryScore",
    "Segment"
))
```

```python
from pyspark.sql.functions import avg
segment_summary = (
    rfm_segmented
    .groupBy("Segment")
    .agg(
        avg("Recency").alias("AvgRecency"),
        avg("Frequency").alias("AvgFrequency"),
        avg("Monetary").alias("AvgMonetary"),
        count("*").alias("CustomerCount")
    )
)
display(segment_summary)
```
