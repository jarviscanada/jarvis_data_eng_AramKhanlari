```python
transactions_df = spark.read.table("bronze.transactions")
users_df = spark.read.table("bronze.users")
cards_df = spark.read.table("bronze.cards")
mcc_df = spark.read.table("bronze.mcc_codes")
fraud_df = spark.read.table("bronze.fraud_labels")
```

```python
from pyspark.sql.functions import  col, lpad, regexp_replace

fraud_df = fraud_df.withColumn(
    "is_fraud",
    col("fraud_label") == "Yes"
).drop("fraud_label")
fraud_df = fraud_df.withColumn("transaction_id",col("transaction_id").cast("int"))

transactions_df = transactions_df.withColumn("zip",lpad(regexp_replace(col("zip").cast("string"), r"\.0$", ""),5,"0"))
```

```python
from pyspark.sql.functions import to_timestamp, dayofweek, hour, weekofyear, col

transactions_silver = (
    transactions_df.alias("t")
    .join(
        fraud_df.alias("f"),
        col("t.id") == col("f.transaction_id"),
        "left"
    )
    .drop("transaction_id")
    .join(
        mcc_df.alias("m"),
        col("t.mcc") == col("m.mcc_code"),
        "left"
    )
    .drop("mcc_code")
    .withColumn("date", to_timestamp(col("date")))
    .withColumn("day_of_week", dayofweek("date"))
    .withColumn("hour", hour("date"))
    .withColumn("week", weekofyear("date"))
)
```

```python
from pyspark.sql.functions import regexp_replace, col, trim
users_silver = (users_df
                .withColumn("yearly_income", regexp_replace("yearly_income", "[$,]", "").cast("double"))
                .withColumn("per_capita_income", regexp_replace("per_capita_income", "[$,]", "").cast("double"))
                .withColumn("total_debt",regexp_replace("total_debt", "[$,]", "").cast("double"))
                .withColumn("gender", trim("gender"))
                .withColumn("address", trim("address"))
)
```

```python
from pyspark.sql.functions import col, lpad
from pyspark.sql.functions import to_date, concat, lit
cards_silver = (cards_df
                .withColumn("expires", to_date(concat(lit("01/"), col("expires")), "dd/MM/yyyy"))
                .withColumn("acct_open_date", to_date(concat(lit("01/"), col("acct_open_date")), "dd/MM/yyyy"))
                .withColumn("has_chip", col("has_chip") == "YES")
                .withColumn("card_on_dark_web", col("card_on_dark_web") == "yes")
                .withColumn("cvv", lpad(col("cvv").cast("string"),3,"0"))
)
```

```python

display(transactions_silver)
display(users_silver)
display(cards_silver)
```

```python
transactions_silver.write.mode("overwrite").saveAsTable("silver.transactions")
users_silver.write.mode("overwrite").saveAsTable("silver.users")
cards_silver.write.mode("overwrite").saveAsTable("silver.cards")
```
