```python
from pyspark.sql.functions import explode, col
mcc_raw = spark.read.option("multiline", "true").json(
    "/Volumes/etl_in_databricks/bronze/tic_volume/mcc_codes.json"
)
#getting data in columns, transform it to rows
row_dict = mcc_raw.collect()[0].asDict()
mcc_list = [(k, v) for k, v in row_dict.items()]

mcc_bronze = spark.createDataFrame(mcc_list, ["mcc_code", "mcc_description"])

display(mcc_bronze)
```

```python
from pyspark.sql.functions import explode, col
from pyspark.sql.types import StructType, MapType, StringType

fraud_schema = StructType().add(
    "target", MapType(StringType(), StringType())
)

fraud_bronze = (
    spark.read
    .schema(fraud_schema)
    .option("multiline", "true")
    .json("/Volumes/etl_in_databricks/bronze/tic_volume/train_fraud_labels.json")
    .select(explode(col("target")).alias("transaction_id", "fraud_label"))
)
```
